## 🧾 Dataset Description — Motor Insurance Churn

This dataset simulates a **portfolio of motor insurance policyholders** from a fictitious insurance company.  
The main objective is to **predict customer churn**, i.e., whether a policyholder is likely to **let their policy lapse or not renew** at the end of the policy term.

The dataset contains **20,000 records**, each representing a customer with their personal, policy, and behavioral information.

---

### 🎯 Objective

The target variable is **`Churn`**:

- `1` → The customer **did not renew** their policy (churned).  
- `0` → The customer **renewed** their policy (retained).

The goal is to **analyze the key factors driving churn** and to **build a machine learning model** that can predict the probability of churn.

---

### 📊 Feature Overview

| Column Name | Description |
|--------------|-------------|
| **Customer_ID** | Unique identifier for each policyholder. |
| **Age** | Age of the customer (in years). |
| **Gender** | Gender of the policyholder (`Male` / `Female`). |
| **Region** | Geographic region where the policy was issued. |
| **Vehicle_Age** | Age of the insured vehicle (in years). |
| **Vehicle_Type** | Type of vehicle (`Sedan`, `SUV`, `Truck`, etc.). |
| **Annual_Premium** | Yearly premium amount paid by the customer (in currency units). |
| **Policy_Tenure** | Duration of the policy with the insurer (in years). |
| **Num_Claims** | Number of insurance claims filed by the customer. |
| **Last_Claim_Amount** | Amount of the most recent claim (0 if none). |
| **Coverage_Type** | Type of insurance coverage (`Basic`, `Extended`, `Premium`). |
| **Satisfaction_Score** | Customer satisfaction rating (scale: 1 = very dissatisfied, 10 = very satisfied). |
| **Num_Interactions** | Number of contacts/interactions with the insurer (calls, emails, etc.). |
| **Renewal_Offer_Rate** | Discount or rate offered during the renewal process. |
| **Payment_Frequency** | Frequency of premium payments (`Monthly`, `Quarterly`, `Annual`). |
| **Churn** | Binary target variable: 1 = churned, 0 = retained. |
